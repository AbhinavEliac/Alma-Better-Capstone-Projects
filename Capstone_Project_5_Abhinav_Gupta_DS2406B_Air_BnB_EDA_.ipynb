# **Project Summary -**

The Exploratory Data Analysis (EDA) of the Airbnb dataset indicates very important information about the distribution of the listings, the effects of the room types, the pricing strategies, and the determinant of whether a business operates or closes. Most of the Airbnb listings, active and closed, are very concentrated in the main neighborhood groups of Brooklyn, Manhattan, and Queens. These markets, though favored by hosts, are also subject to the greatest competition. Most importantly, almost 30 to 40 percent of companies providing Entire home/apts and Private room products have closed shop in Manhattan and Brooklyn. This large closure rate indicates oversaturated markets in which incorrect pricing, subpar service, and insufficient differentiation could be leading to business failure. Conversely, shared room postings, although fewer in number, exhibit a considerably lower closure rate, reflecting an established niche market or less competitive environments.

Detailed visualizations such as Box plots and Heatmaps also emphasize the price trends that distinguish successful postings from those that close. For instance, in Manhattan, the median price per night for open Entire home/apt listings is around \$200, while for closed businesses it drops to \$175. A similar pattern is seen in Brooklyn, where open listings in this category average \$150 per night compared to \$130 for closed ones. This suggests that listings able to justify a higher price through better amenities or service quality tend to perform better. In most neighborhoods, except Staten Island, open listings always report greater price per night values, confirming that consumers are ready to pay more when they feel value. Staten Island is different; there, lower prices work best towards business success, indicating price sensitivity among guests is greater within the area. Listings with fewer minimum night requirements also perform well, particularly for group rooms. Numerous closed shared room listings also had longer minimum stays, suggesting that decreasing guest flexibility may be one of the contributing factors to closures.

More Statistical Analysis with Scatter plots and Correlation matrices reveals patterns in the spatial and relational data that explain business performance. Neighborhoods such as Murray Hill and Kips Bay in Manhattan, and East Flatbush and Flushing in Brooklyn and Queens, respectively, have more open businesses and proportionally fewer closures. This suggests these markets might be more conducive to new listings, perhaps because there is more balanced demand and competition. Conversely, Williamsburg in Brooklyn and Midtown Manhattan each have more closed listings than open listings, which should be a warning to new hosts opening up those markets. Correlation analysis reveals high positive correlations between total cost and price and between total cost and minimum nights, indicating that hosts who overprice their listings without value delivery can run higher total costs and fewer bookings. There is also data indicating that customers are more price conscious with private and shared rooms, especially in non central boroughs, and either demand competitive pricing or quality service or both.

Overall, this EDA provides a complete picture of the New York City Airbnb business environment. It emphasizes that selection of room type, price setting, availability, and neighborhood choices are all essential to the profitability of a listing. For hosts and investors, these findings can inform strategic actions such as neighborhood targeting, room type selections, price competitiveness, and service quality maintenance to reduce the closure of businesses and maximize long term expansion.
