# **Project Summary -**

Booking.com is a platform, through which people engage in Hotel Bookings. This platform is currently struggling with High Cancelations, and is in urgent need of Data Driven Solutons, so that it is able to improve booking efficiency, predict guest needs, and tailor their offerings to better meet market demands, which will help to streamline operations, maximize occupancy rates, and deliver exceptional service, ensuring a competitive edge in the dynamic hospitality industry.

The platform is Dominated by the City Hotel Booking Requests, which also see a greater percentage of Cancelations, as compared to the Resort Hotel Booking Requests. Online TA mode is most Popular medium of Bookings, which also give high amount of Cancelations as well, indicating that there is an urgent need to take corrective measures in the Working of Online Travel Agents. Transient Customers Dominate the platform, and also give rise to high cancelations, indicating poor Customer Management.

Despite the Yearly increase in the Customers with Booking Requests, there has been an increase in the cancelation rates as well, indicating a Overcrowding and Poor Customer Management.

The Reasons why this has happened are because of Higher Lead Times, Lower Number Of Special Requests, Lower Number of Booking Changes, No Deposit Type Payment and Marginally because of Higher Cost in case of Resort Hotel Booking.

The Best Part about this Dataset is the fairly negligible amount of Customers have Days Spent in Waiting List, which shows greater Hotel Availability. And, the Worst Part about this Dataset is that Customers with Previous Cancelations, return less, and even when they do, the same Customers have very high Cancelation Rates, indicating Customer frustration due to lack of improvement in the Business's handling of Customers.

Interesting Insights come into play when we separate Customers into Adults and Adults with Children and Babies category, indicating the seasonal nature of bookings and greater proportion of Resort Hotel Booking Requests, when Kids are involved. But, due to lack lower number of Special Requests, the Customers accompanying Kids, tend to Cancel more.

This brings us to the Conclusion that, Booking.com needs to Urgently look into the Online Travel Agent activities and streamline them through effective administrative measures, Lower the Booking Prices, promote Special Requests through advertisements, offers or Discounts, make it Customer friendly to make Changes in the Bookings, and change the Payment Deposit Type to Refundable and Non-Refundable, instead of No Deposit Type before Booking Confirmation.

Employing these Measures will surely boost the Business and this Summarises the Booking.com Exploratory Data Analysis Project.
