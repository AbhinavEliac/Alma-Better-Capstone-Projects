# **Project Summary -**

The Exploratory Data Analysis of the Playstore data gave useful insights using various kinds of charts like bar charts, box plots, pie charts, heatmaps, histograms, and pairplots. Bar chart analysis showed that the Family and Game categories dominate both in app number and installations, particularly in the free app category. These results reflect the developers' interest and the areas of audience interest, and hence it reflects a highly competitive level. In the same way, paid apps in the two categories reflect high usage, meaning people are ready to pay for these segments. This makes them good options for monetization through paid models.

Box plots revealed that across many categories, especially in the Family and Game categories, average application ratings are above 4, a good indicator of user satisfaction. Surprisingly, more sparsely populated categories like Health & Fitness, Art & Design, and Books & Reference have high ratings and respectable install volumes, which are reflections of untapped potential and potential for developers seeking a competitive edge. These niches' ratings suggest that high-quality applications can succeed even in scenarios where overall application counts are low. Second, the combination of healthy ratings and moderate installation volumes suggests an extremely high degree of engagement and loyalty on the part of the users, making these environments supportive of innovative application development.

Pie charts and heatmaps provided a close examination of ratings of content and user interests. "Everyone" is the most common rating in every category, which indicates a diversified audience, while Teen-rated content has gained growing traction in Social and Game categories. The findings also discovered that, with fewer applications in some genres, user interest (as indicated by installs) is still strong. The heatmaps showed that highly rated apps are concentrated in the 4.3 to 4.5 rating range, indicating a competitive marketplace where quality is the key. Also, in the paid segment, Family and Game apps that fell under the "Everyone 10+" and "Mature 17+" content ratings had strong installations, indicating that product design needs to be aligned with target age groups to be made more successful.

Lastly, sentiment analysis with histograms and pairplots offered a qualitative viewpoint. Sentiment polarities of most are slightly positive, and subjectivity levels of 50–60% reveal that reviews are intrinsically authentic. High sentiment polarity in both Family and Communication categories reflect greater user satisfaction, while Games, despite high volume, offer room for improvement. Pairplots also revealed substantial trends like declining installs with increasing app price, and greater app sizes are accompanied by fewer ratings. The correlation heatmaps showed a moderate correlation between installs and reviews, affirming that user feedback tends to grow with popularity. Overall, these findings in combination provide actionable insights to app developers and companies to target appropriate categories, price optimally, and improve quality to fuel growth and revenue.

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Play Store Data

# Dropping duplicates
ps_df_psd_sample.drop_duplicates(inplace = True)

ps_df_psd_sample
````

````python
# Defining Datatype Modification Function for Installs, Reviews and Price Columns
def dt_mod_func(col):

  # Storing Each Cell's Characters in a List of Strings
  x = list(str(col))
  num_str = ''
  p = 1

  # Here the Numeric Values will be taken out from the and the Alphabets will be used for Unit Conversion with 'p'
  for t in x:

    try:

      y = int(t)
      num_str = num_str + str(y)

    except ValueError:

      if str(t) == '.':

        y = t
        num_str = num_str + str(y)

      elif str(t) in 'mkKM':

        if t in 'kK':

          p = 1000

        elif t in 'mM':

          p = 1000000

      else:

        continue

  # Converting the resulting String into a Numeric Type and using the Appropriate Type Conversion on it
  return float(num_str) * p if num_str else 0.0


# Defining Datatype Modification Function for Size Column
def size_mod_func(size):

  # Storing Each Cell's Characters in a List of Strings
  s = list(str(size))
  num_size = ''
  q = 1

  # Here the Numeric Values will be taken out from the and the Alphabets will be used for Unit Conversion with 'q'
  for u in s:

    try:

      r = int(u)
      num_size = num_size + str(r)

    except ValueError:

      if str(u) == '.':

        r = u
        num_size = num_size + str(r)

      elif str(u) in 'mkgsMKGS':

        if u in 'kK':

          q = 1 / 1024
          break

        elif u in 'gG':

          q = 1024
          break

        elif u in 'mM':

          q = 1
          break

      else:

        continue

  # Converting the resulting String into a Numeric Type and using the Appropriate Type Conversion on it
  return float(num_size) * q if num_size else 0.0


# Applying the Datatype modification Functions
ps_df_psd_sample['Installs'] = ps_df_psd_sample['Installs'].apply(dt_mod_func)

ps_df_psd_sample['Reviews'] = ps_df_psd_sample['Reviews'].apply(dt_mod_func)

ps_df_psd_sample['Price'] = ps_df_psd_sample['Price'].apply(dt_mod_func)

ps_df_psd_sample['Size MBs'] = ps_df_psd_sample['Size'].apply(size_mod_func)


# Moving the Size MBs Column
column_to_move_2 = ps_df_psd_sample.pop('Size MBs')
ps_df_psd_sample.insert(5, 'Size MBs', column_to_move_2)



# Converting 'Last Updated' to datetime
ps_df_psd_sample['Last Updated'] = pd.to_datetime(ps_df_psd_sample['Last Updated'], errors = 'coerce')

# Spliting the 'Genres' column into 'Genre' and 'Sub Genre'
ps_df_psd_sample['Genre'] = ps_df_psd_sample['Genres'].apply(lambda x: x.split(';')[0])
ps_df_psd_sample['Sub Genre'] = ps_df_psd_sample['Genres'].apply(lambda x: x.split(';')[1] if len(x.split(';')) > 1 else 'No Sub genre')

ps_df_psd_sample.info()
````

````python
# Defining Version Normalisation Function in the form of only 'Major_number.Minor_number'
def extract_version(version_string):

    if version_string == 'Varies with device':

        return version_string

    # Using regex to find Major and Minor version numbers
    match = re.match(r'(\d+(\.\d+){0,1})', str(version_string))

    if match:

        return match.group(1)

    # Handling cases that don't match the expected pattern
    return 'Unknown'

# Applying the function to 'Current Ver' and 'Android Ver'
ps_df_psd_sample['Current Ver'] = ps_df_psd_sample['Current Ver'].apply(extract_version)
ps_df_psd_sample['Android Ver'] = ps_df_psd_sample['Android Ver'].apply(extract_version)

ps_df_psd_sample.info()
````

````python
# Finding Median Rating of all Categories and Storing them as a Dictionary
med_rev_df = ps_df_psd_sample.groupby(['Category'])['Rating'].median()
dict_1 = med_rev_df.to_dict()

# Defining the Function to Fill the Null Ratings with Medians of their Categories
def rev_fill_1 (category, rating):

  if np.isnan(rating):

    rating = dict_1[category]

  else:

    rating = rating

  return rating

# Applying the Function
ps_df_psd_sample['fill_rating'] = ps_df_psd_sample.apply(lambda x: rev_fill_1(x['Category'], x['Rating']), axis = 1)

ps_df_psd_sample.drop(columns = ['Rating'], inplace = True)

ps_df_psd_sample['Rating'] = ps_df_psd_sample['fill_rating']
ps_df_psd_sample.drop(columns = ['fill_rating'], inplace = True)

ps_df_psd_sample.info()
````

````python
# Finding Median Sentiment_Polarity of all Apps and Storing them as a Dictionary
med_sent_pol_df = ps_df_ur_sample.groupby(['App'])['Sentiment_Polarity'].median()
dict_2 = med_sent_pol_df.to_dict()

# Defining the Function to Fill the Null Sentiment_Polarity with Medians of their Apps
def rev_fill_2 (app, sent_pol):

  if np.isnan(sent_pol):

    sent_pol = dict_2[app]

  else:

    sent_pol = sent_pol

  return sent_pol

# Applying the Function
ps_df_ur_sample['fill_sent_pol'] = ps_df_ur_sample.apply(lambda x: rev_fill_2(x['App'], x['Sentiment_Polarity']), axis = 1)

ps_df_ur_sample.drop(columns = ['Sentiment_Polarity'], inplace = True)

ps_df_ur_sample['Sentiment_Polarity'] = ps_df_ur_sample['fill_sent_pol']
ps_df_ur_sample.drop(columns = ['fill_sent_pol'], inplace = True)

ps_df_ur_sample.info()
````

````python
# Finding Median Sentiment_Subjectivity of all Apps and Storing them as a Dictionary
med_sent_sub_df = ps_df_ur_sample.groupby(['App'])['Sentiment_Subjectivity'].median()
dict_3 = med_sent_sub_df.to_dict()

# Defining the Function to Fill the Null Sentiment_Subjectivity with Medians of their Apps
def rev_fill_3 (app, sent_sub):

  if np.isnan(sent_sub):

    sent_sub = dict_3[app]

  else:

    sent_sub = sent_sub

  return sent_sub

# Applying the Function
ps_df_ur_sample['fill_sent_sub'] = ps_df_ur_sample.apply(lambda x: rev_fill_3(x['App'], x['Sentiment_Subjectivity']), axis = 1)

ps_df_ur_sample.drop(columns = ['Sentiment_Subjectivity'], inplace = True)

ps_df_ur_sample['Sentiment_Subjectivity'] = ps_df_ur_sample['fill_sent_sub']
ps_df_ur_sample.drop(columns = ['fill_sent_sub'], inplace = True)

ps_df_ur_sample.info()
````

````python
# Defining Function to Fill Null Sentiments according to the Signs of their Sentiment Polarities
def rev_fill_4 (app, sent, sent_pol):

  if pd.isnull(sent):

    if sent_pol > 0:

      sent = 'Positive'

    elif sent_pol == 0:

      sent = 'Neutral'

    else:

      sent_pol = 'Negative'

  else:

    sent = sent

  return sent

# Applying the Function
ps_df_ur_sample['fill_sent'] = ps_df_ur_sample.apply(lambda x: rev_fill_4(x['App'], x['Sentiment'], x['Sentiment_Polarity']), axis = 1)

ps_df_ur_sample.drop(columns = ['Sentiment'], inplace = True)

ps_df_ur_sample['Sentiment'] = ps_df_ur_sample['fill_sent']
ps_df_ur_sample.drop(columns = ['fill_sent'], inplace = True)

ps_df_ur_sample.info()
````

#### Chart - 2 - Bivariate Bar Chart - Total Installs Per Category in Powers of 10

````python
# Chart - 2 visualization code

# Grouping data by powers of 10
ps_df_psd_1_cat_pop_app_inst_df = ps_df_psd_1_cat_pop_app_inst.reset_index()
ps_df_psd_1_cat_pop_app_inst_df['Installs_log10'] = np.log10(ps_df_psd_1_cat_pop_app_inst_df['Installs'])
ps_df_psd_1_cat_pop_app_inst_df['Installs_power_of_10'] = np.floor(ps_df_psd_1_cat_pop_app_inst_df['Installs_log10'])

# Replacing infinite values with 0
ps_df_psd_1_cat_pop_app_inst_df['Installs_power_of_10'] = ps_df_psd_1_cat_pop_app_inst_df['Installs_power_of_10'].replace([np.inf, -np.inf], 0)

# Getting unique powers of 10
powers_of_10 = sorted(ps_df_psd_1_cat_pop_app_inst_df['Installs_power_of_10'].unique())

# Creating subplots for each power of 10
fig, axes = plt.subplots(nrows=len(powers_of_10), figsize=(8, 4 * len(powers_of_10)))

for i, power in enumerate(powers_of_10):

    subset_df = ps_df_psd_1_cat_pop_app_inst_df[ps_df_psd_1_cat_pop_app_inst_df['Installs_power_of_10'] == power]
    subset_df = subset_df.sort_values('Installs', ascending = False)

    ax = axes[i] if len(powers_of_10) > 1 else axes
    subset_df.plot(kind = 'bar', x = 'Category', y = 'Installs', ax = ax, legend = False)
    ax.set_title(f'Total Installs per Category (10^{int(power)} to 10^{int(power)+1})')
    ax.set_xlabel('Category')
    ax.set_ylabel('Total Installs')
    ax.tick_params(axis = 'x', rotation = 90)

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

Bar charts effectively compare categorical data by illustrating discrete values, facilitating quick visual recognition of differences in magnitude among categories.

##### 2. What is/are the insight(s) found from the chart?

With App Installs in the terms of 10^11, the Categories which Dominate the App installs are Game, Communication, Social, Productivity, Tools and Family.

Also the Games with 3 \* 10^10 Installs have the highest installs, and Family with 1 \* 10^10 have the lowest Installs, even lower than Social Category with 1.5 \* 10^10 Installs, which had around 300 Apps to it's name.

##### 3. Will the gained insights help creating a positive business impact?
